| 用法 | 說明 |
|:---:|:---:|
| `range(m)` | 走訪 0 到 m-1 的數字 |
| `range(n, m)` | 走訪 n 到 m-1 的數字  |
| `range(n, m, step)` | 以step為間距，走訪 n 到 m-1 的數字 |

##### for

```python
# Ctrl + D 選擇相同的文字做編輯，使用完按 Esc 退出
n = 0
for char in "contents":
    print(char, end="") # end 的預設是"\n"，會換行
    n += 1
    print(n, end=", ")
```

```python
# range(m-1)
for i in range(5):
    print(i)
```

```python
# range(n, m-1)
for i in range(1, 11):
    print(i, end = ", ")
```

```python
# range(n, m-1, step)
for i in range(1, 11, 2):
    print(i, end = ", ")
```

### 小試身手 for  
請產出 a 開頭的 n 個偶數(含 a 本身)  
Ex:  
a = 5, n = 3  
Output : 6 8 10

a = 4, n = 5  
Output : 4 6 8 10 12

```python
a = 4
n = 5
if a % 2 == 0:
    a = a
else:
    a += 1

for i in range(n):
    print(a)
    a += 2
```

```python
# 嘗試產出
"""
* * * * *
 * * * *
  * * *
   * *
    *
"""
n = 5
blank = 0
while n > 0:
    for i in range(blank):
            print(" ", end = "")
    for i in range(n):
        print("*", end =" ")
    print(end="\n")
    n -= 1
    blank +=1

n = 5
for i in range(n):
     print(" " * i + ("*"+" ") * (n-i))


# 嘗試產出
"""
    *
   * *
  * * *
 * * * *
* * * * *
"""
n = 1
blank = 4
while n <= 5:
    for i in range(blank):
            print(" ", end = "")
    for i in range(n):
        print("*", end =" ")
    print(end="\n")
    n += 1
    blank -=1

n = 5
for i in range(n+1):
     print(" " * (n-i) + ("*"+" ") * (i))
```

##### while

```python
count = 1

while count <= 5:
    print(count, end=", ")
    count += 1

print("count 已經不小於 5，結束迴圈")
print(count)
```
