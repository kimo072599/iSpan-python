條件語句 if-elif-else

```python
'''
條件判斷
if ...
if ... else ...
if ... elif ... else ...

'''
```

##### if

```python
num = 20
if num > 10:
    print("num 大於 10")
```

```python
a = 10
b = 5
if a != b:
    print("a 不等於 b")
```

```python
score = 50
passing_score = 60

if score < passing_score:
    score += 10
    print("加分")

# Tab 縮排
# SHIFT + Tab 反縮排
print("學生的分數是", score)
```

##### if...else...

```python
num = 5
if num > 10:
    print("num 大於 10")
else:
    print("num 沒有大於 10")
```

##### if...elif... (else)

```python
num = 10
print(type(num))
```

```python
num = 6
if num > 10:
    print("num 大於 10")
elif num % 2 == 0:
    print("num 是偶數")
elif type(num) == int:
    print("num 資料型態為int")
else:
    print("num 沒有大於 10")

# if...elif 有順序性，須注意
# else 不一定必要
```
