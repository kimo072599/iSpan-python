```python
# 類別名稱
class Car:
    # 類別屬性 (Class Attribute)
    wheel = 4
    # 建構函示 (是一個 magic method)
    def __init__(self, color, brand, seats = 4):
        # 實例屬性
        self.color = color
        self.brand = brand
        self.seats = seats
    # 方法
    def info(self):
        print(f"Color: {self.color}, brand: {self.brand}, wheel: {self.wheel}")

# 建立物件
car1 = Car("red", "Toyota", seats = 6)
# 取得物件的實例屬性
print(car1.color)
print(car1.brand)
print(car1.seats)
# 取得物件的類別屬性
print(car1.wheel)

car2 = Car("black", "Benz")
print(car2.color)
print(car2.brand)
print(car2.seats) # 建立物件時沒有指定，就帶入預設實例屬性
print(car2.wheel) # 類別屬性是共用的


# 使用類別方法
car1.info()
car2.info()
```

實作：將編輯檔案的功能寫成 Class

```python
class EditTxt: # class 命名使用駝峰式命名
    def __init__(self, file_path):
        self.file_path = file_path

    def read_file(self):
        with open(self.file_path, "r", encoding="utf-8") as f:
            doc = f.read()
            return doc
    def insert_msg(self, msg):
        with open(self.file_path, "a", encoding="utf-8") as f:
            f.write(msg)

txt_file = EditTxt("../Module2/2-1.py")
print(txt_file.read_file())

file2 = EditTxt("../Module9/9-1.py")
file2_str = file2.read_file()
file2.insert_msg("# 測試 method 使用")
print(file2_str)
```
