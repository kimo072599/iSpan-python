##### Class 設計

```python
class EditTxt: 
    def __init__(self, file_path):
        self.file_path = file_path

    def read_file(self):
        with open(self.file_path, "r", encoding="utf-8") as f:
            doc = f.read()
            return doc
    def insert_msg(self, msg):
        with open(self.file_path, "a", encoding="utf-8") as f:
            f.write(msg)
```

```python
class child(EditTxt):
    def insert_msg(self, msg):
        with open(self.file_path, "a", encoding="utf-8") as f:
            f.write(msg+"\n")
```
