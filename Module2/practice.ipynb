百貨公司周年慶 消費金額超過 2000 要折 200  

請撰寫一支程式  
計算客戶消費的結帳金額

```python
spending = int(input("請輸入消費金額："))
print("消費金額為：", spending)
if spending >= 2000:
    spending -= 200
print("結帳金額為：", spending)
```

```python
# 滿 4000 折 400，滿 6000 折 600，以此類推
spending = int(input("請輸入消費金額："))
print("消費金額為：", spending)
orispending = spending
discount = 0
while spending >= 2000:
    spending -= 2000
    discount += 200
print("結帳金額為：", orispending - discount)
```

```python
spending = int(input("請輸入消費金額："))
print("消費金額為：", spending)
discount = (spending // 2000) * 200
print("結帳金額為：", spending - discount)
```
