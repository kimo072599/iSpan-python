|基本型態|創建方式|
|:---:|:---:|
| int | `num = 10` |
| float | `num = 2.5` |
| boolean | `parse_ok = True` |
| string | `name = "Kimo"` |

|容器型態|創建方式|
|:---:|:---:|
|list| `my_list = [1, 2, 3, 4]` |
|tuple|`my_tuple = (1, 2, 3)`|
|set| `my_set = {1, 2, 3}` |
|dict| `my_dict = {"name":"Kimo", "age":27}` |


整數 (int)、浮點數 (float)

```python
num1 = 10
num2 = 2.5
print(num1)
print(num2)
```

布林值 (boolean)

```python
parse_ok = True
parse_no = False
print(parse_ok)
print(parse_no)
```

字串 (string)

```python
message = "Hello World!"
print(message)
```

```python
data = "True"
parse_ok = True
print(data)
print(parse_ok)

# type() 變數型別的檢查
print(type(data))
print(type(parse_ok))
```

列表 (list)

```python
num_list =[1, 2, 3, 4, 1]
print(type(num_list), num_list)
```

元組 (tuple)

```python
num_tuple = (1, 2, 3, 1)
print(type(num_tuple), num_tuple)
```

集合 (set)

```python
num_set = {1, 2, 3, 1}
print(type(num_set), num_set)
```

字典 (dict)

```python
user_info = {"name":"Kimo", "age":27}
print(type(user_info), user_info)
```

```python
data_list =[]
print(type(data_list), data_list)
# 創建一個空的 dict
x = {}
print(type(x), x)
```

```python
# 創建一個空的 set
empty_set = set()
print(type(empty_set), empty_set)

empty_set2 = ()
print(type(empty_set2), empty_set2)
```
