身分運算子 is / is not

```python
x = 10
y = 10
print(x)
print(y)
print(id(x))
print(id(y))
```

```python
if x is y:
    print("x 和 y 參照到同一個記憶體位置")
```

```python
if x == y:
    print("x 等於 y ")
```

```python
x = [1, 2, 3]
y = [1, 2, 3]

if x == y:
    print("x 等於 y ")

print(id(x))
print(id(y))
if x is y:
    print("x 和 y 參照到同一個記憶體位置")
```

```python
x = [1, 2, 3]
y = x

if x == y:
    print("x 等於 y ")

print(id(x))
print(id(y))
if x is y:
    print("x 和 y 參照到同一個記憶體位置")
```
