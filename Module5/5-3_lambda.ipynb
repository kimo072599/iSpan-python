### lambda 匿名函式  
用法  

`function_name = lambda arguments: expression`

````python
# 建立函式
def add_num(x, y):
    return x + y
print(add_num(1, 5))
````

````python
# 基本用法
add_num_lambda = lambda x, y: x + y
print(add_num_lambda(1, 5))
````

````python
# 條件判斷
def check_even(num):
    if num % 2 == 0:
        return True
    else:
        return False
    
print(check_even(5))
````

````python
check_even_lambda = lambda num: True if num % 2 == 0 else False
print(check_even_lambda(5))
````

### 小試身手  
請建立一個 list ，並在該 list 內產生多個 `[1, 2, x]` 的 list (x 的值要遞增)  
x 到 10 即可  
Ex :  
```Python
[ [1, 2, 1], [1, 2, 2], [1, 2, 3]...[1, 2, 10] ]
```

````python
outside = []
for i in range(1, 11):
    outside.append([1, 2, i])

print(outside)
````

每三個 list 中間的值要改成 5  
Ex:  
```python
[ [1, 2, 1], [1, 2, 2], [1, 5, 3], [1, 2, 4], [1, 2, 5], [1, 5, 6]...]
```

````python
outside = []
for i in range(1, 11):
    if i % 3 ==0:
        outside.append([1, 5, i])
    else:
        outside.append([1, 2, i])

print(outside)

outside.sort()
print(outside)
# 預設會依 index 順序做排序

outside.sort(key=lambda a : a[2])
print(outside)
````

````python
# 如果不用 lambda 怎麼寫出上面的要求?
outside = []
for i in range(1, 11):
    if i % 3 ==0:
        outside.append([1, 5, i])
    else:
        outside.append([1, 2, i])


def sublist_third_index(sublist):
    return sublist[2]

outside.sort(key=sublist_third_index)
print(outside)
````
