```python
# 全域變數
name ="global name"

# 定義函式
def greeting():
    msg = "Hello"
    name = "local name"
    return msg +" "+ name
# 執行函式
data = greeting() 
# function 內的變數名稱與全域變數衝突時
# 呼叫 funtion 時會以 funtion 內部的變數為主
# 但不會改變全域變數
print(data)
# 輸出 name
print(name)
```

##### 使用 global 讓函式存取全域變數的值

```python
# 全域變數
name ="global name"

# 定義函式
def greeting():
    msg = "Hello"
    global name # 呼叫全域變數
    name = "local name"
    return msg +" "+ name

# 執行函式
data = greeting() 
print(data)
# 輸出 name
print(name)
```
