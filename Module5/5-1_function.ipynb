```python
# 建立 function 2：function 的 return 值
def greeting():
    msg = "Hello World!"
    str_list = msg.split()
    return str_list

greeting()
# 這個呼叫並不會打印出東西，因為函式中的程式碼不會打印東西
# 如果要看到這個函式的值，就要用一個變數來接住函式的 return 值
x = greeting()
print(x)

# 如果呼叫程式時沒有加()
y = greeting
print(y)
# 會顯示該值為一個 function

print(greeting())
print(greeting)
```

```python
# 建立 function 3：設定 function 的參數
def greeting(msg):
    str_list = msg.split()
    return str_list

data = greeting("Hello world")
print(data)

data = greeting()
# 呼叫時若不填入參數，就會報錯
print(data)
```

```python
# 建立 function 3-1：可設定多個參數
def greeting(msg: str, name):
    str_list = msg.split()
    str_list.append(name)
    return str_list

data = greeting("Hello World!", "Kimo")
print(data)
data = greeting(name = "Kimo", msg = "Hello World!")
# 用 = 的方法填入參數，可不依照順序
print(data)
```

```python
# 建立 function 4：設定 function 參數的預設值
def greeting(msg = "Good Morning!"):
    str_list = msg.split()
    return str_list

data = greeting()
print(data)

data = greeting("Hello world")
print(data)
```

```python
# 建立 function 4-2：給參數的預設值的順序
# 若有參數不給預設值，必須放在有預設值的參數前
def greeting(name, msg = "Good Morning!"):
    str_list = msg.split()
    str_list.append(name)
    return str_list

data = greeting("user")
print(data)

data = greeting("user","Hello world")
print(data)
```

```python
# 定義參數時若把沒有預設值的參數放在有預設的後面就會報錯
def greeting(msg = "Good Morning!", name):
    str_list = msg.split()
    str_list.append(name)
    return str_list
```

```python
# 改寫 Module2 practice
def get_discount(spending):
    discount = (spending // 2000) * 200
    #print("結帳金額為：", spending - discount)
    return spending - discount
print(get_discount(6000))
```

```python
def get_discount(total, discount):
    discount = (total // 2000) * 200
    #print("結帳金額為：", total - discount)
    return total - discount
data = [
    {
        "total": 6000,
        "discount": 200
    },
    {
        "total": 8000,
        "discount": 200

    }
]

for d in data:
    print(d["total"],d["discount"])
    answer = get_discount(d["total"],d["discount"])
    print(f"折扣後的金額： {answer}元")
```

```python
def get_list_odd_num(num_list):
    odd_num_list = []
    for i in num_list:
        if i % 2 == 1:
            odd_num_list.append(i)
    return odd_num_list

print(get_list_odd_num([i for i in range(50)]))
```

解釋 *args 和 *keargs

```python
def add_nums(*args, **kwargs):
    print(args)
    print(kwargs)
    sum = 0
    for i in args:
        sum += i

    for key, value in kwargs.items():
        sum += value

    return sum
ans = add_nums(1, 2, 3, 4, 5, n1=100, n2=200)
print(ans)
```
