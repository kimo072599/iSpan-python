```python
# 初始化 dict
# key 也可以放數字，但實務上通常會有更明確的命名
fruit_price_dict = {"apple": 100, "orange": 50, "tomato": 30}
print(fruit_price_dict)
```

```python
# 印出 蘋果的價格
fruit_price_dict = {"apple": 100, "orange": 50, "tomato": 30}
print(fruit_price_dict["apple"])

# 須注意：當呼叫不存在的 key，程式就會報錯
print(fruit_price_dict["banana"])
```

```python
# 修改 橘子的價格
fruit_price_dict = {"apple": 100, "orange": 50, "tomato": 30}
fruit_price_dict["orange"] = 70
print(fruit_price_dict)
```

```python
# 刪除 番茄
fruit_price_dict = {"apple": 100, "orange": 50, "tomato": 30}
del fruit_price_dict["tomato"]
print(fruit_price_dict)
```

```python
fruit_price_dict = {"apple": 100, "orange": 50, "tomato": 30}
fruit_price_dict.pop("tomato")
print(fruit_price_dict)
```

```python
# 當 key-value 很多組，應適當排版
fruit_price_dict = {
        "apple": 100,
        "orange": 50,
        "tomato": 30
    }
```

```python
# 建立一個 value 含有 dict 的 dict
user_info = {
    "name": "Kimo",
    "age": 27,
    "info": {
        "nickname": "kimo",
        "favorite_role": "ironman",
        "phone":[
            "0912345678",
            "0987654321"
        ]
    }
}
```

```python
# 取得 nickname
print(user_info["info"]["nickname"])
```

```python
# 取得所有 phone_number
print(user_info["info"]["phone"])
for phone_number in user_info["info"]["phone"]:
    print(phone_number)
```

```python
'''
改寫 fruit_dict 每個水果需要 price, quantity/amount, color
apple 有 綠色 和 紅色
orange 有 橘色
tomato 有 紅色
'''
fruit_dict = {
    "apple":{
        "price": 100,
        "amount": 10,
        "color": {"green", "red"} #{}?,[]?
    },
    "orange":{
        "price": 50,
        "amount": 6,
        "color": {"orange"} # 因為上面建立了 set，這邊最好就也給相同的資料結構
    },
    "tomato":{
        "price": 30,
        "amount": 23,
        "color": {"red"}
    }
}

print(fruit_dict)
```

##### dict 的合併

```python
# dict 的合併
# dict.update()
fruit_dict2 = {
    "kiwi":{
        "price": 66,
        "amount": 78,
        "color": {"green"}
    }
}

fruit_dict.update(fruit_dict2)
print(fruit_dict)
```

```python
# 特殊用法：修改 value 的其他作法
# 用 ** 將一個大 dict 塞進新的 dict
# ** 的概念：將 dict 的 key-value 拆解，放進新的 dict 中
fruit_dict3 = {
    **fruit_dict,
    "kiwi":{
        "price": 66,
        "amount": 78,
        "color": {"yellow"}
    }
}
# 實務上再寫的時候並不會用新的 dict 來接
# 而是直接用原本的 dict，將其覆蓋
print(fruit_dict3)
```

##### 另外的 print 方式 `pprint` (pretty-print) 

```python
# 從 Module 引入 function
# 打印出來的資料不會按原順序，但也不影響原資料
from pprint import pprint
pprint(fruit_dict3)
```

```python
'''
新增顏色
apple 有新的品種 顏色要有黃色
tomato 有綠色
'''
fruit_dict3["apple"]["color"].add("yellow")
fruit_dict3["tomato"]["color"].add("green")
pprint(fruit_dict3)
```
