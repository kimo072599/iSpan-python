```python
# 移除指定的"值"
# list.remove(<value>)
ids = [1, 2, 3, 4, 5, 6]
ids.remove(3)
print(ids)
```

```python
ids = [1, 2, 3, 4, 5, 6]
for _ in range(5):
    ids.insert(1, 9)
print(ids)
```

```python
ids.remove(9)
# 只會移除第一個 9 
print(ids)
```

```python
# 嘗試搭配 for 迴圈來移除所有 9
for i in ids:
    if i == 9:
        ids.remove(9)
print(ids)
# 輸出 [1, 9, 9, 2, 3, 4, 5, 6]
# 因為 value 位置會隨著移除而改變，所以會有跳過的現象
```

```python
# 嘗試搭配 for 迴圈來移除所有 9
for i in ids:
    if i == 9:
        ids.pop(9)
print(ids)
# 輸出 [1, 9, 9, 2, 3, 4, 5, 6]
# 因為 value 位置會隨著移除而改變，所以會有跳過的現象
```

```python
ids = [1, 2, 3, 4, 5, 6]
for _ in range(5):
    ids.insert(1, 9)
print(ids)
ids_check = ids
for i in ids_check:
    if i == 9:
        ids.remove(9)
print(ids)
```

```python
# 利用 while 搭配成員運算子
while 9 in ids:
    ids.remove(9)
print(ids)
```

```python
ids = [1, 2, 3, 4, 5, 6]
for _ in range(5):
    ids.append(10)
print(ids)
# 計算 list 內的某個值 共有幾個
# list.count(<value>)
cnt_10 = ids.count(10)
print(cnt_10)
```

List 的合併

```python
# for 迴圈的寫法，當 list 很大，就會吃執行時間
ids1 = [1, 2 ,3 ,4]
ids2 = [5, 6, 7]

for i in ids2:
    ids1.append(i)
print(ids1)
```

```python
# list.extend(<list2>)
ids1 = [1, 2 ,3 ,4]
ids2 = [5, 6, 7]
ids1.extend(ids2)
print(ids1)
```

```python
# 也可以直接相加，缺點是要花費額外記憶體處存相加的值
ids1 = [1, 2 ,3 ,4]
ids2 = [5, 6, 7]

total_ids = ids1 + ids2
print(total_ids)
```

```python
# 不要用成 append
# 雖然可以執行，但會變成在最後掛進一個 list
# list 資料型別混用會很麻煩
ids1 = [1, 2 ,3 ,4]
ids2 = [5, 6, 7]
ids1.append(ids2)
print(ids1)
```

### list 搭配 for 迴圈

```python
# 建立一個 1~10 的 list
num_list = []
for i in range(10):
    num_list.append(i+1)
print(num_list)
```

```python
# 建立一個 1~50 的 list，只要偶數
num_list = []
for i in range(1, 51):
    if i % 2 == 0:
        num_list.append(i)
print(num_list)
```

建立一個 list，裡面要放遞增的 list

```python
# 建立一個 list，裡面要放遞增的 list
# 例如 [[1], [1, 2], [1, 2, 3]...[1, 2, ...10]]
num_list = []
final_list = []
for i in range(1, 11):
    num_list.append(i)
    final_list.append(num_list)
print(final_list)
# 這樣寫不會達成預期，因為最後 final_list 裡面就是十個 num_list
# 然後再參照最後的 num_list 記憶體位置的值，印出十個 1 ~ 10 的 list

outside = []
for i in range(1, 11):
    inside = []
    for c in range(1, i+1):
        inside.append(c)
    outside.append(inside)
print(outside)
```

```python
num_list = []
final_list = []
for i in range(1, 11):
    for num in range(1, i+1):
        num_list.append(num)
    final_list.append(num_list)
print(final_list)
print(num_list)
# 如果把 num_list 的空值定在外面，num_list 的建立會出問題
# 而且最後還是十個一模一樣的 num_list
```

```python
# 每一次的賦值，就會有一個記憶體位置
inside = []
print(id(inside))
inside = []
print(id(inside))
inside = []
print(id(inside))
```

```python
# .copy() 的使用
num_list = []
final_list = []
for i in range(10):
    num_list.append(i+1)
    final_list.append(num_list.copy())
print(final_list)
```

```python
# enumerate() 同時取得 index 與 value
ids = [2, 50, 3, 72, 25]
for i in enumerate(ids):
    print(i)


ids = [2, 50, 3, 72, 25]
for index, value in enumerate(ids):
    print(f"index: {index}, value: {value}")
```

### List comprehension 列表推導式
`new_list = [expression for item in iterable [if condition]]`  
  
`list的名稱 = [表達式 for 元素 in 迭代對象 [if 判斷式(...的條件)]]`

```python
# 建立一個 1~10 的 list
num_list = [i for i in range(1, 11)]
print(num_list)
```

```python
# 建立一個 1~10 的二次方的 list
square_list = [i**2 for i in range(1, 11)]
print(square_list)
```

```python
# 建立一個 list，裡面要放遞增的 list
# 例如 [[1], [1, 2], [1, 2, 3]...[1, 2, ...10]]
outside = []
for i in range(1, 11):
    inside = []
    for c in range(1, i+1):
        inside.append(c)
    outside.append(inside)
print(outside)

# inside = [c for c in range(1, i+1)]
outside = [[c for c in range(1, i+1)] for i in range(1, 11)]
print(outside)
```

```python
# 建立一個 1~50 的 list，只要偶數
num_list = []
for i in range(1, 51):
    if i % 2 == 0:
        num_list.append(i)
print(num_list)

num_list = [i for i in range(1, 51) if i % 2 == 0]
print(num_list)
```
