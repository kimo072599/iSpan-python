|string方法| 功能 |
|:---:|:---:|
|`str.replace()` | 替換字串 |
|`str.strip()` | 去除兩側空格 |
|`str.upper()` | 字串小寫變大寫 |
|`str.lower()` | 字串大寫變小寫 |

```python
# 字串變數初始化
name = "Kimo"
print(name)
print(name[0])

# name[0] = "D"，是不可行的
```

```python
# 替換字串
# "方法"：定義在 class 底下的 function
# 方法也是 function 的一種(但 function 不一定是方法)
'''
str.replace()
將 string 中的 str1 替換成 str2
'''

name = "Kimo"
new_name = name.replace('K', "D")
print(new_name)
```

```python
# 去除兩側空格
'''
str.strip()
去除字串 string 左、右兩邊的空格
'''
data = "   |__  data  __|   "
print(data)

new_data = data.strip() # 只能去除兩端空格
print(new_data)

# 如果要清除字串間的空格，就要使用 .replace
no_space_data = data.replace(" ","")
print(no_space_data)
```

```python
# 去除單邊
'''
str.lstrip
str.rstrip
'''
data = "1   |__  data  __|   1"
new_data = data.lstrip("1")
print(new_data)
```

```python
# 字串小寫變大寫
'''
str.lower()
將字串 string 裡的字母全部改成小寫
'''
data = "CAR"
new_data = data.lower()
print(new_data)

'''
str.upper()
將字串 string 裡的字母全部改成大寫
'''
print("good".upper())
# 前提是該 function 要有回傳直，才能直接在 print 裡面使用
# 例如 list 的 append function 就沒有回傳直，不能直接寫在 print 中
```

```python
# 字串分割成 list
# str.split()
greeting = "hello world! hello John"
str_list = greeting.split(" ")
print(str_list)
```

```python
# 如果還想要細分到每個字元，就沒辦法
# 只能用 for 迴圈去拆
greeting = "hello world! hello John"
str_list = greeting.split(" ")
for data_str in str_list:
    for char in data_str:
        print(char, end = ", ")
    print()
```
