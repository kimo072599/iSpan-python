```python
# tuple 初始化: 第一種
data = ("Big", "Data", "is", "good")
print(data)
print(type(data))
```

```python
# tuple 初始化: 第二種 -- 直接不括號
data = "Kimo", 27
print(data)
print(type(data))

name, age ="Kimo", 27
print(name)
print(age)
```

```python
data = "Kimo", 27, "AIPE02"
print(data)

name, age, _class = data
print(name)
print(age)
print(_class)
```

```python
# 透過指定索引輸出值
data = "Kimo", 27, "AIPE02"
print(data[0])
```

```python
# 複數變數修改值
# Jarvis：沒有意義，不教
```

```python
# tuple 不可以修改指定索引的值
data[0] = 100
```

```python
# (list)可以修改指定索引的值
num_list = [1, 2, 3]
num_list[0] = 100
print(num_list)
```

```python
# 用 for 迴圈逐一輸出資料
data = "Kimo", 27, "AIPE02"
for i in data:
    print(i)
```

```python
# 用 len 計算 tuple 資料個數
data = "Kimo", 27, "AIPE02"
print(data)
print(len(data))
```
