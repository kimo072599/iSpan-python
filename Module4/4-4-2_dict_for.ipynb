```python
fruit_dict = {    
    'apple': {
        'price': 100,
        'quantity': 50,
        'color': ['green', 'red']
    },
    'orange': {
        'price': 50,
        'quantity': 30,
        'color': 'orange' # 不建議的寫法
    }
}
print(fruit_dict)
```

```python
# 使用 for 迴圈讀取 dict 資料1 
# 只會印出 key 值
for key in fruit_dict:
    print(key)
```

```python
# 使用 for 迴圈讀取 dict 資料2
# 若想取得 value，使用 .values()
for key in fruit_dict.values():
    print(key)
```

```python
# 使用 for 迴圈讀取 dict 資料3
# 若想要 key 也想要 value，使用 .items()
for key, value in fruit_dict.items():
    print(key)
    print(value)
```

```python
for key, value in fruit_dict.items():
    #print(value["color"])
    for color in value["color"]:
        print(color)
```

```python
# 取得 dict 的 value，並避免出錯
# print(fruit_dict["kiwi"]) 會導致報錯
# 使用 .get()
print(fruit_dict.get("apple"))
print(fruit_dict.get("kiwi")) # 沒有 key 值回傳 none
```

```python
# 資料處理
fruit_color_list = [
    {'apple': 'red'}, 
    {'grape': 'purple'}, 
    {'kiwi': 'yellow'}, 
    {'apple': 'green'}, 
    {'kiwi': 'green'}, 
    {'grape': 'green'}]
```

```python
# 將資料整理成 {'apple': ['red', 'green'], 'grape': ['purple', 'green'], 'kiwi': ['yellow', 'green']}
sorted_fruit_color_dicts = {}
for fruit_color_dict in fruit_color_list:
    for fruit, color in fruit_color_dict.items():
        if sorted_fruit_color_dicts.get(fruit) == None:
            sorted_fruit_color_dicts[fruit] = [color]
        else:
            sorted_fruit_color_dicts[fruit].append(color)
        
print(sorted_fruit_color_dicts)
```
