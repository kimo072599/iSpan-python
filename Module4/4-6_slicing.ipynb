##### slice 透過 [ : ] 進行分割
`[n:m]` 走訪 n 到 m - 1 的元素

```python
# 初始化一個字串
data = "Hello,World!"
print(data)
print(data[0])
print(data[6])
```

```python
# 從頭取到 5（不包含第 5 個元素，實際為 0 ~ 4）
data = "Hello,World!"
print(data[0:5])
```

```python
# 從頭取到 5（不包含第 5 個元素，實際為 0 ~ 4）
data = "Hello,World!"
print(data[:5])
```

```python
# 從 6 取到尾
data = "Hello,World!"
print(data[6:])
```

```python
# 從 7 取到 9（不包含第 9 個元素，實際為 7 ~ 8）
data = "Hello,World!"
print(data[7:9])
```

##### 使用負號

```python
# 從倒數第 10 取到倒數第 7（不包含倒數第 7 的元素）
data = "Hello,World!"
print(data[-10:-7])
```

```python
# 從倒數第 6 取到尾
data = "Hello,World!"
print(data[-6:])
```

```python
# 從頭取到倒數第 7（不包含倒數第 7 的元素）
data = "Hello,World!"
print(data[:-7])
```

```python
# 也可以使用在 list
num_list = [i for i in range(1, 11)]
print(num_list)
print(num_list[-1])
print(num_list[-3: ])
print(num_list[: 3])
```

```python
# 實做練習 將每個單字改成大寫開頭
string04 = "hello world! hello john"
string04_list = string04.split(" ")
new_str = ""
for word in string04_list:
    new_str += word[0].upper() + word[1:] + " "
print(new_str)
```

```python
# 實做練習 將每個單字改成大寫開頭
string04 = "hello world! hello john"
string04_list = string04.split(" ")
new_str = ""
for word in string04_list:
    new_str += word.capitalize() + " "
print(new_str)
# 這個方式最後會多一個空格，要多一個步驟去除空格
new_str = new_str.strip()
print(new_str)
```

```python
# str.join() 的用法
str_list = ["Hello", "World"]
msg = "_".join(str_list)
print(msg)
```

```python
string04 = "hello world! hello john"
string04_list = string04.split(" ")
print(string04_list)
new_str = " ".join([i[0].upper() + i[1:] for i in string04_list])
print(new_str)
```
